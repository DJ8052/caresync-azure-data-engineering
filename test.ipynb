```sql
SELECT
    COUNT(*) AS total_rows,
    COUNT(DISTINCT patient_id) AS unique_patients,
    COUNT(*) - COUNT(DISTINCT patient_id) AS duplicate_count
FROM dbw_caresync_dev.silver.patients;
```

```sql
SELECT *
FROM dbw_caresync_dev.silver.hospitals
WHERE hospital_name = 'Riverside Specialty Hospital';
```

```sql
SELECT
    d.doctor_id,
    d.first_name,
    d.last_name,
    d.specialty,
    h.hospital_name
FROM dbw_caresync_dev.silver.doctors d
JOIN dbw_caresync_dev.silver.hospitals h
    ON d.hospital_id = h.hospital_id
WHERE h.hospital_name = 'Riverside Specialty Hospital';
```

```sql
SELECT
    patient_id,
    first_name,
    city,
    state,
    registered_hospital_id,
    updated_at
FROM dbw_caresync_dev.silver.patients
WHERE patient_id IN (1, 2500);
```
